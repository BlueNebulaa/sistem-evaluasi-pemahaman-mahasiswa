EXTRACT PDF

```python
import fitz
import os


def extract_pdf(pdf_path, output_path):
    doc = fitz.open(pdf_path)

    all_text = []

    for page_number, page in enumerate(doc, start=1):
        text = page.get_text("text")

        if text.strip():
            all_text.append(
                f"\n===== HALAMAN {page_number} =====\n{text}"
            )

    final_text = "\n".join(all_text)

    with open(output_path, "w", encoding="utf-8") as f:
        f.write(final_text)

    print(f"Berhasil mengekstrak: {pdf_path}")
    print(f"Jumlah halaman: {len(doc)}")
    print(f"Output: {output_path}")
```

```python
def read_file(path):
    if not os.path.exists(path):
        raise FileNotFoundError(f"File tidak ditemukan:\n{path}")
    with open(path, "r", encoding="utf-8") as f:
        return f.read()

def clean_content(text):
    text = text.replace("\r\n", "\n").replace("\r", "\n")
    text = re.sub(r"===== HALAMAN \d+ =====", "", text, flags=re.IGNORECASE)
    text = re.sub(r"(?m)^\s*\d+\s*$", "", text)
    text = re.sub(r"(?m)^\s*PEDOMAN ETIKA KECERDASAN ARTIFISIAL\s*$", "", text, flags=re.IGNORECASE)
    text = re.sub(r"(?m)^\s*POKJA 3-ETIKA\s*$", "", text, flags=re.IGNORECASE)
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()
```

```python
def normalize_heading(text):
    text = text.lower()
    text = text.replace("kecerdasan artifisial", "kecerdasan artifisial")
    text = re.sub(r"[^a-z0-9]+", " ", text)
    return " ".join(text.split())
```

```python
def find_modul_heading_lines(lines):
    found = []
    patterns = [
        (r"^\s*A\.\s+Deskripsi Umum Modul", "A"),
        (r"^\s*A1\.\s*Capaian Pelatihan", "A1"),
        (r"^\s*A\.1\.\s*Tujuan Pelatihan", "A.1"),
        (r"^\s*A\.2\.\s*Indikator Capaian Pelatihan", "A.2"),
        (r"^\s*A\.3\.\s*Pokok Bahasan", "A.3"),
        (r"^\s*A\.4\.\s*Alur Pelatihan", "A.4"),
        (r"^\s*B\.?\s*Kecerdasan Artifisial Generatif", "B"),
        (r"^\s*B\.1\.?\s*Ruang Lingkup", "B.1"),
        (r"^\s*B\.2\.?\s*Prinsip Kerja", "B.2"),
        (r"^\s*B\.3\.?\s*Kualitas Data", "B.3"),
        (r"^\s*B\.4\.?\s*Implementasi KA Generatif", "B.4"),
        (r"^\s*B\.5\.?\s*Contoh Aktivitas", "B.5"),
        (r"^\s*C\.?\s*Etika dan Resiko", "C"),
        (r"^\s*C\.1\.?\s*Etika dalam", "C.1"),
        (r"^\s*C\.1\.1\.?\s*Transparansi", "C.1.1"),
        (r"^\s*C\.1\.2\.?\s*Hak Cipta", "C.1.2"),
        (r"^\s*C\.2\.?\s*Resiko", "C.2"),
        (r"^\s*C\.2\.1\.?\s*Potensi Penyalahgunaan", "C.2.1"),
        (r"^\s*C\.2\.2\.?\s*Bias", "C.2.2"),
        (r"^\s*C\.2\.3\.?\s*Pelanggaran Privasi", "C.2.3"),
        (r"^\s*C\.3\.?\s*Teknologi Deepfake", "C.3"),
        (r"^\s*C\.3\.1\.?\s*Contoh Aktivitas", "C.3.1"),
        (r"^\s*D\.?\s*Daftar Pustaka", "D")
    ]
    compiled = [(re.compile(pattern, re.IGNORECASE), code) for pattern, code in patterns]

    for i, line in enumerate(lines):
        clean = line.strip()
        if not clean:
            continue
        for pattern, code in compiled:
            if pattern.search(clean):
                found.append({"line": i, "code": code})
                break
    return found
```

```python
def find_pedoman_heading_lines(lines):
    found = []
    patterns = [
        (r"^\s*A\.\s*PANCASILA SEBAGAI PEDOMAN ETIKA", "A"),
        (r"^\s*a\.\s*Pendahuluan", "A.a"),
        (r"^\s*b\.\s*Pengertian Etika", "A.b"),
        (r"^\s*c\.\s*Kecerdasan Artifisial", "A.c"),
        (r"^\s*d\.\s*Mewujudkan Kecerdasan Artifisial", "A.d"),
        (r"^\s*B\.\s*PEDOMAN ETIKA", "B"),
        (r"^\s*a\.\s*Konsep Manusia sebagai", "B.a"),
        (r"^\s*b\.\s*Keamanan Teknis", "B.b"),
        (r"^\s*c\.\s*Tata Kelola Data", "B.c"),
        (r"^\s*d\.\s*Transparansi, Akuntabilitas", "B.d"),
        (r"^\s*e\.\s*Pengembangan nilai", "B.e"),
        (r"^\s*f\.\s*Mencegah dan menghindari", "B.f"),
        (r"^\s*g\.\s*Non-Diskriminatif", "B.g"),
        (r"^\s*h\.\s*Kewaspadaan", "B.h"),
        (r"^\s*C\.\s*PRINSIP IMPLEMENTASI", "C"),
        (r"^\s*a\.\s*Andal", "C.a"),
        (r"^\s*b\.\s*Terbuka dan responsif", "C.b"),
        (r"^\s*c\.\s*Kejelasan Pertanggungjawaban", "C.c"),
        (r"^\s*d\.\s*Auditable", "C.d"),
        (r"^\s*D\.\s*PERAN KOLABORASI", "D"),
        (r"^\s*A\.\s*Peran Pemerintah", "D.2.A"),
        (r"^\s*B\.\s*Peran Universitas", "D.2.B"),
        (r"^\s*C\.\s*Peran Industri", "D.2.C"),
        (r"^\s*D\.\s*Peran Komunitas", "D.2.D"),
        (r"^\s*E\.\s*Peran Komite Etika", "D.2.E"),
        (r"^\s*E\.\s*CHECKLIST KECERDASAN", "E")
    ]
    compiled = [(re.compile(pattern, re.IGNORECASE), code) for pattern, code in patterns]

    for i, line in enumerate(lines):
        clean = line.strip()
        if not clean:
            continue
        for pattern, code in compiled:
            if pattern.search(clean):
                found.append({"line": i, "code": code})
                break
    return found
```

```python
def make_chunks(text, document, headings, section_info):
    lines = text.splitlines()
    chunks = []

    for i, heading in enumerate(headings):
        start = heading["line"]
        end = headings[i + 1]["line"] if i + 1 < len(headings) else len(lines)
        content = clean_content("\n".join(lines[start:end]))

        if not content:
            continue

        code = heading["code"]
        title = section_info.get(code, code)

        chunk = {
            "chunk_id": f"{document}_{len(chunks)+1:02d}",
            "document": document,
            "section": code,
            "title": title,
            "source_line_start": start + 1,
            "source_line_end": end,
            "content": content
        }
        chunks.append(chunk)

    return chunks
```

```python
def create_pedoman_chunks(text):
    lines = text.splitlines()
    headings = find_pedoman_heading_lines(lines)

    checklist_index = None
    for h in headings:
        if h["code"] == "E":
            checklist_index = h["line"]
            break

    if checklist_index is not None:
        extra_patterns = [
            (r"^\s*Data\s*$", "E.a"),
            (r"^\s*Algoritma\s*$", "E.b"),
            (r"^\s*Model\s*$", "E.c")
        ]
        compiled = [(re.compile(pattern, re.IGNORECASE), code) for pattern, code in extra_patterns]

        for i in range(checklist_index + 1, len(lines)):
            clean = lines[i].strip()
            for pattern, code in compiled:
                if pattern.match(clean):
                    headings.append({"line": i, "code": code})
                    break

        model_heading = next((h for h in headings if h["code"] == "E.c"), None)

        if model_heading:
            for i in range(model_heading["line"] + 1, len(lines)):
                clean = lines[i].lower()
                if "mitigasi" in clean or "litbangjirap" in clean or "checklist" in clean:
                    headings.append({"line": i, "code": "E.d"})
                    break

    headings = sorted(headings, key=lambda x: x["line"])

    section_info = {code: title for code, title in PEDOMAN_HEADINGS}
    section_info.update({
        "E.a": "Data",
        "E.b": "Algoritma",
        "E.c": "Model",
        "E.d": "Matriks Usulan Upaya Mitigasi Risiko di Setiap Tahapan Litbangjirap"
    })

    return make_chunks(
        text=text,
        document="pedoman_etika",
        headings=headings,
        section_info=section_info
    )
```

```python
modul_text = read_file(MODUL_FILE)
modul_headings = find_modul_heading_lines(modul_text.splitlines())
modul_info = {code: title for code, title in MODUL_HEADINGS}
modul_chunks = make_chunks(
    text=modul_text,
    document="modul_etika_ai",
    headings=modul_headings,
    section_info=modul_info
)

# 12. PROSES PEDOMAN
pedoman_text = read_file(PEDOMAN_FILE)
pedoman_chunks = create_pedoman_chunks(pedoman_text)

# 13. GABUNGKAN
all_chunks = modul_chunks + pedoman_chunks
```

```python
def save_jsonl(chunks, filename):
    path = os.path.join(OUTPUT_DIR, filename)
    with open(path, "w", encoding="utf-8") as f:
        for chunk in chunks:
            f.write(json.dumps(chunk, ensure_ascii=False) + "\n")
    return path

modul_output = save_jsonl(modul_chunks, "modul_etika_ai_chunks.jsonl")
pedoman_output = save_jsonl(pedoman_chunks, "pedoman_etika_chunks.jsonl")
all_output = save_jsonl(all_chunks, "all_etika_chunks.jsonl")
```

LOAD TO QDRANT (EMBED & LOAD)

```python
import os
import json
from dotenv import load_dotenv
from fastembed import TextEmbedding
from qdrant_client import QdrantClient
from qdrant_client.models import Distance, VectorParams, PointStruct

load_dotenv()

INPUT_FILE = r"C:\Users\RENDRA\OneDrive\Documents\prosada (RAG)\NEW\chunking_etika\all_etika_chunks.jsonl"
COLLECTION_NAME = "interview_knowledge"
EMBEDDING_MODEL = "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
VECTOR_SIZE = 384

QDRANT_URL = os.getenv("cluster_endpoint")
QDRANT_API_KEY = os.getenv("api_qdrant_interview")
```

```python
client = QdrantClient(url=QDRANT_URL, api_key=QDRANT_API_KEY)
print("Berhasil terhubung ke Qdrant Cloud")

chunks = []
with open(INPUT_FILE, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if line:
            chunks.append(json.loads(line))

print(f"Jumlah chunk: {len(chunks)}")
```

```python
print("Loading embedding model...")
embedding_model = TextEmbedding(model_name=EMBEDDING_MODEL)
print("Embedding model berhasil dimuat.")

existing_collections = [collection.name for collection in client.get_collections().collections]

if COLLECTION_NAME not in existing_collections:
    client.create_collection(
        collection_name=COLLECTION_NAME,
        vectors_config=VectorParams(size=VECTOR_SIZE, distance=Distance.COSINE)
    )
    print(f"Collection '{COLLECTION_NAME}' berhasil dibuat.")
else:
    print(f"Collection '{COLLECTION_NAME}' sudah tersedia.")
```

```python
texts = [chunk["content"] for chunk in chunks]

embeddings = list(embedding_model.embed(texts))
print(f"{len(embeddings)} vector")

points = []
for index, (chunk, embedding) in enumerate(zip(chunks, embeddings)):
    point = PointStruct(
        id=index + 1,
        vector=embedding.tolist(),
        payload={
            "chunk_id": chunk["chunk_id"],
            "document": chunk["document"],
            "section": chunk["section"],
            "title": chunk["title"],
            "source_line_start": chunk["source_line_start"],
            "source_line_end": chunk["source_line_end"],
            "content": chunk["content"]
        }
    )
    points.append(point)

print("Mengupload vector ke Qdrant Cloud...")
client.upsert(collection_name=COLLECTION_NAME, points=points)

print("UPLOAD BERHASIL")
print(f"Collection : {COLLECTION_NAME}")
print(f"Jumlah     : {len(points)} vector")
print(f"Dimensi    : {VECTOR_SIZE}")

collection_info = client.get_collection(COLLECTION_NAME)
print(f"Jumlah vector di Qdrant: {collection_info.points_count}")
```
